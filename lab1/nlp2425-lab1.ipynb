````python
# Importamos Stanza, cliente de acceso a la librería CoreNLP
import stanza

# Si no se ha hecho previamente, descargamos el modelo de CoreNLP preentrenado en inglés
stanza.download('en') # Descomentar solo la primera vez
````

````python
# Cargamos el modelo y lo encapsulamos en un pipeline de procesadores de lenguaje natural:
# - tokenize, mwt: sentence and word tokenizers
# - pos: part-of-speech tagger
# - lemma: lemmatizer
# - constituency, depparse: constituency and dependency parsers
# - sentiment: sentiment analyzer
# - ner: name entity recognizer
nlp = stanza.Pipeline('en')
````

````python
# Establecemos el texto a analizar: una reseña sobre un teléfono móvil
review = """
        The phone is okay, but it doesn't meet all my expectations. 
        Its camera is not as good as I expected; in low light, the photos come out blurry. 
        The price is too high for the features it offers. What disappoints me the most is the battery, which doesn't last long. 
        Its design, on the other hand, is quite nice and comfortable to use."
        """
````

***Etiquetado gramatical - Part-of-Speech (PoS) tagging***

````python
# Preprocesamos la review (si se considera necesario)
#review = review.lower()

# Procesar la review con Stanza (cliente CoreNLP)
doc = nlp(review)

for sentence in doc.sentences:
    print(f"Sentence: '{sentence.text}'")
````

````python
# Imprimimos por pantalla tablas con los etiquetados gramaticales de las oraciones de la reseña
for sentence in doc.sentences:
    print("\n*** PoS tags ***")
    print(f"Sentence: '{sentence.text}'")
    
    print(f"{'ID':<5} {'Word':<15} {'PoS Tag':<15} {'XPOS Tag':<15}")
    print("-" * 50)
    
    for word in sentence.words:
        print(f"{word.id:<5} {word.text:<15} {word.upos:<15} {word.xpos:<15}")
````

***PREGUNTA 1***. ¿Cuál es el significado y posibles valores de word.upos y word.xpos?

*RESPUESTAS*:


`word.upos` muestra el pos universal de una palabra.Según el enlace https://universaldependencies.org/u/pos/index.html, hay 17 tipos de upos que son ADJ,ADP,ADV,AUX,CCONJ,DET,INTJ,NOUN,NUM,PART,PRON,PROPN,PUNCT,SCONJ,SYM,VERB y X.

`word.xpos` significa el Penn Treebank (PTB) tag de una palabra.Hay 45 tipos, por ejemplo: 
* Nouns: `NN` = singular, `NNS` = plural, `NNP` = singular proper, `NNPS` = plural proper
* Verbs: `VB` = base, `VBD` = past tense, `VBG` = gerund, `VBP/VPZ` = (non) 3sg person, `MD` = modal
* Adjectives: `JJ` = base, `JJR` = comparative, `JJS` = superlative
* Adverbs: `RB` = base, `RBR` = comparative, `RBS` = superlative
* Prepositions & subordinating conjunctions: `IN`
* Pronouns: `PRP` = personal, `PRP$` = possessive
* Determiners: `DT`
* Coordinating conjunction: `CC`

***Análisis de constituyentes - Constituency parsing***

````python
# Función recursiva para imprimir por pantalla un árbol de constituyentes
def print_constituency_tree(constituency, level):
    node_text = constituency.label
    children = constituency.children
    
    indent = "  " * level
    print(f"{indent}{node_text}")
    
    for child in children:
        if isinstance(child, str):
            print(f"{indent}  {child}")
        else:
            print_constituency_tree(child, level + 1)

# Imprimimos por pantalla el análisis de constituyentes de las oraciones de la reseña
for sentence in doc.sentences:
    print("\n*** Constituency parse ***")
    print(f"Sentence: '{sentence.text}'")

    constituency = sentence.constituency
    print(f"Parse (bracket notation):\n{constituency}")
    
    # Ahora, imprimimos la estructura en formato de taxonomía o directorio
    print("Parse (tree format):")
    print_constituency_tree(constituency, 0)
````

***PREGUNTA 3***. ¿Qué significan las relaciones 'nsubj', 'amod', 'advmod' y 'conj'? Al definir cada relación, incluye ejemplos (word, deprel, head) de las tablas anteriores en las que la relación aparece.

1. `nsubj — nominal subject (sujeto nominal)`
Indica el sujeto nominal (sintáctico) de una cláusula. Suele ser el elemento que realiza la acción del verbo principal o el tema en construcciones copulativas (con cop, como "is", "was").

    Ejemplos de las tablas:
    - phone (NOUN), nsubj, okay (ADJ)
    - it (PRON), nsubj, meet (VERB)
    - camera (NOUN), nsubj, good (ADJ)
    - price (NOUN), nsubj, high (ADJ)
    - What (PRON), nsubj, battery (NOUN)
    - design (NOUN), nsubj, nice (ADJ)


2. `amod — adjectival modifier (modificador adjetival)`
Un adjetivo (o frase adjetival) que modifica directamente a un sustantivo o pronombre.

    Ejemplos de las tablas:
    - low (ADJ), amod, light (NOUN)   ("low light")
    - other (ADJ), amod, hand (NOUN)  ("the other hand")

3. `advmod — adverbial modifier (modificador adverbial)`
Un adverbio (o frase adverbial) que modifica a un verbo, adjetivo, otro adverbio o a toda la cláusula. Incluye negaciones (n't), grados (too, quite), duración (long), etc.
    
    Ejemplos de las tablas:
    - n't (PART), advmod, meet (VERB)
    - too (ADV), advmod, high (ADJ)
    - quite (ADV), advmod, nice (ADJ)
    - long (ADV), advmod, last (VERB)

4. `conj — conjunct (elemento coordinado)`
Une elementos coordinados (palabras, frases o cláusulas) mediante una conjunción coordinante (and, but, or…). El primer elemento coordinado actúa como cabeza, y los siguientes se conectan a él con conj. La conjunción coordinante se etiqueta como cc.

    Ejemplos de las tablas:
    - meet (VERB), conj, okay (ADJ)   ("okay, but it doesn't meet…" — coordinación con "but")
    - comfortable (ADJ), conj, nice (ADJ)   ("nice and comfortable")

### Extracción de opiniones sobre aspectos (de un producto) en una reseña.

````python
def extract_aspect_opinion_pairs(review):
    doc = nlp(review)
    pairs = []

    for sentence in doc.sentences:
        for word in sentence.words:
            head_word = sentence.words[word.head - 1] if word.head > 0 else None
            head_pos = head_word.upos if head_word else 'None'
            
            if word.upos in {"NOUN", "PROPN"}:
                if head_word and head_word.upos == "ADJ":
                    pair = (word.text, head_word.text)
                    if pair not in pairs:
                        pairs.append(pair)
                        print(f"1. {pair}")

            if word.upos == "ADJ":
                if head_word and head_word.upos in {"NOUN", "PROPN"}:
                    pair = (word.text, head_word.text)
                    if pair not in pairs:
                        pairs.append(pair)
                        print(f"2. {pair}")

    return pairs

print(review)
pairs = extract_aspect_opinion_pairs(review)
#print(pairs)
````

````python
def extract_aspect_opinion_pairs(review):
    doc = nlp(review)
    pairs = set()

    for sentence in doc.sentences:
        words = sentence.words

        for word in words:
            if word.head == 0:
                continue
            head = words[word.head - 1]

            aspect = None
            opinion = None

            # amod
            if word.deprel == "amod" and word.upos == "ADJ" and head.upos in {"NOUN", "PROPN"}:
                aspect = head.text.lower()
                opinion = word.text.lower()

            # copulativas
            if word.deprel == "cop" and word.upos == "AUX":
                verb_idx = word.head
                adj = words[verb_idx - 1]
                if adj.upos == "ADJ":
                    opinion = adj.text.lower()
                    for subj in words:
                        if subj.head == verb_idx and subj.deprel == "nsubj":
                            aspect = subj.text.lower()
                            break

            # xcomp
            if word.deprel == "xcomp" and word.upos == "ADJ":
                for subj in words:
                    if subj.head == word.head and subj.deprel == "nsubj":
                        aspect = subj.text.lower()
                        opinion = word.text.lower()
                        break

            # conj adj
            if word.deprel == "conj" and word.upos == "ADJ" and head.upos == "ADJ":
                opinion = word.text.lower()
                for subj in words:
                    if subj.head == head.head and subj.deprel == "nsubj":
                        aspect = subj.text.lower()
                        break

            if aspect and opinion:
                pairs.add((aspect, opinion))

    print("Pares extraídos (aspecto, opinión):")
    for i, (a, o) in enumerate(sorted(pairs), 1):
        print(f"{i}. ('{a}', '{o}')")

    return list(pairs)

pairs = extract_aspect_opinion_pairs(review)
````
